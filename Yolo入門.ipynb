```python
#他のオプション
#　boxes→画像の検知して、結果を表示するボークス※結果を見たらわかりやすい・Bounding box
# mask→セグメンテーションマスク結果用マスクオブジェクト
# keypoints→
results = model("/content/drive/MyDrive/test_data/input/bus.jpg",tracker="bytetrack.yml")
for resultn in results:
    boxes = resultn.boxes
    resultn.plot(labels=True,conf=True)
    print(boxes)

# orig_shape: (1080, 810)→H*W（高さ＊幅）画像サイズ
# shape: torch.Size([5, 6])→　５Bounding box・５つ物を検知した、６→（x1,y1,x2,y2,自身値,クラスID）

# 出力
# boxes.cls→tensor([5., 0., 0., 0., 0.])　：：　検知されたオブジェクトのCoCO　クラスインデクス→５つ
# boxes.conf→tensor([0.8806, 0.8720, 0.8608, 0.8462, 0.6562])　：：　各検知した物の自身値
# ※画像に検知した物は以下の通りです：
# ４人、一台バス、４人の中の一人の手だけがあるので0.6562は自身値です。
# boxes.data→5*6行列演算子された値
```

#### plot→結果にどういう引数を設定したい・描く
<h4>plot引数
<p>
plot
(conf: bool = True, line_width: float | None = None, font_size: float | None = None, font: str = "Arial.ttf", pil: bool = False, img: ndarray | Tensor | None = None, kpt_radius: int = 5, kpt_line: bool = True, labels: bool = True, boxes: bool = True, masks: bool = True, probs: bool = True, show: bool = False, save: bool = False, filename: str | None = None, color_mode: str = "class", txt_color: tuple[int, int, int] = (255, 255, 255)) -> ndarray
</p>
</br>
<p>
主に使われる引数

* conf→自身値
* labels→検知された物体の種別名を表示する
* txt_color→BBOXの色。</br>
例：txt_color=RGB(0,0,0)→黒</br>
    txt_color=RGB(255,255,255)→白い
</p>

<h3>
TrackとPredict
</h3>
<p>
Yolo predict：このメソッドは生物を検知しながら人間なら人間と他の動物ならその動物名をBBOXで表示します。Predict：予想
</br>
Yolo track：は生物を検知と追跡をします。
tracker引数は適切なモデルを設定したら、一つ物体（移動せずな）を独自ポイントとして、計算しながら生物名と生物独自IDを設定します。</br>
今回使ったtracker：bytetrack
※
